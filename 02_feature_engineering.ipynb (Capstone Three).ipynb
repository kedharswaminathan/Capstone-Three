Notebook Objective

The purpose of this notebook is to prepare the Telco Customer Churn dataset for machine learning.

Based on the findings from the data-understanding notebook, the data requires several transformations before it can be used to train classification models. These transformations include converting TotalCharges to a numerical variable, removing the unique customer identifier, encoding categorical variables, separating the target variable, splitting the data into training and testing sets, and standardizing numerical features.

The goal is to produce clean and consistently formatted training and testing datasets that can be used in the modeling stage.

1. Import Libraries

```python
import pandas as pd 
import numpy as np 

from sklearn.model_selection import train_test_split 
from sklearn.preprocessing import StandardScaler
```

2. Load the Dataset

```python
df = pd.read_csv( r"C:\Users\kedha\Downloads\WA_Fn-UseC_-Telco-Customer-Churn.csv" )

df.head()
```

3. Review the Dataset

3.1 Dataset Shape

```python
print(f"Number of rows: {df.shape[0]}") 
print(f"Number of columns: {df.shape[1]}")
```

3.2 Review Data Types

```python
df.dtypes
```

4. Convert TotalCharges to Numeric

```python
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)
```

```python
df["TotalCharges"].dtype
```

4.1 Check Missing Values

```python
df["TotalCharges"].isnull().sum()
```

```python
df[df["TotalCharges"].isnull()]
```

5. Handle Missing Total Charges

```python
df = df.dropna(subset=["TotalCharges"])
```

```python
df["TotalCharges"].isnull().sum()
```

```python
df.shape
```

6. Remove the Customer ID

```python
df = df.drop(columns=["customerID"])
```

```python
df.head()
```

7. Separate Features and Target

7.1: Define X

```python
X = df.drop(columns=["Churn"])
```

7.2: Define Y

```python
y = df["Churn"]
```

```python
print(f"Features shape: {X.shape}") 
print(f"Target shape: {y.shape}")
```

8. Convert the Target Variable

```python
y = y.map({ "No": 0, "Yes": 1 })
```

```python
y.value_counts()
```

```python
y.value_counts(normalize=True) * 100
```

9. Identify Numerical and Categorical Features

9.1 Numerical Features

```python
numerical_features = X.select_dtypes( include=["int64", "float64"] ).columns.tolist() 
numerical_features
```

9.2 Categorical Features

```python
categorical_features = X.select_dtypes( include=["object"] ).columns.tolist() 
categorical_features
```

10. Create Dummy Variables

```python
X = pd.get_dummies(
    X,
    columns=categorical_features,
    drop_first=True,
    dtype=int
)
```

```python
X.head()
```

```python
print(f"Number of features after encoding: {X.shape[1]}")
```

```python
X.columns.tolist()
```

11. Verify Feature Values

```python
X.dtypes.value_counts()
```

```python
X.isnull().sum().sum()
```

12. Train/Test Split

```python
X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.20, random_state=42, stratify=y )
```

12.1 Review Split Sizes

```python
print(f"Testing features: {X_test.shape}") 
print(f"Testing features: {X_test.shape}")

print(f"Training target: {y_train.shape}") 
print(f"Testing target: {y_test.shape}")
```

12.2 Check Target Distribution

```python
print("Training Churn Distribution:")
print(y_train.value_counts(normalize=True) * 100) 

print("\nTesting Churn Distribution:") 
print(y_test.value_counts(normalize=True) * 100)
```

13. Standardize Numerical Features

```python
scaler = StandardScaler()
```

```python
X_train_scaled = X_train.copy()
X_test_scaled = X_test.copy()

X_train_scaled[numerical_features] = scaler.fit_transform(
    X_train[numerical_features]
)

X_test_scaled[numerical_features] = scaler.transform(
    X_test[numerical_features]
)
```

14. Verify Standardization

```python
X_train_scaled[numerical_features].mean()
```

```python
X_train_scaled[numerical_features].std()
```

15. Final Feature Engineering Review

```python
X_train_scaled.head()
```

```python
X_test_scaled.head()
```

```python
print("Final Training Data:") 
print(X_train_scaled.shape)

print("\nFinal Testing Data:") 
print(X_test_scaled.shape)
```

16. Final Data Quality Checks

16.1 Missing Values

```python
print( "Missing values in training data:", X_train_scaled.isnull().sum().sum() ) 
print( "Missing values in testing data:", X_test_scaled.isnull().sum().sum() )
```

16.2 Infinite Values

```python
print( "Infinite values in training data:", np.isinf(X_train_scaled).sum().sum() ) 
print( "Infinite values in testing data:", np.isinf(X_test_scaled).sum().sum() )
```

16.3 Target Values

```python
print("Training target values:") 
print(y_train.value_counts())

print("\nTesting target values:") 
print(y_test.value_counts())
```

17. Feature Engineering Summary

17.1 Transformations Completed

TotalCharges was converted from a string to a numerical variable.

Records with missing TotalCharges values were removed.

customerID was removed because it is a unique identifier and does not provide predictive information.

The Churn target was converted from categorical values (Yes/No) to binary values (1/0).

Categorical features were converted into dummy variables using one-hot encoding.

The dataset was divided into training and testing subsets using an 80/20 split.

The split was stratified to preserve the distribution of churn in both datasets.

Numerical features were standardized using StandardScaler.

The scaler was fitted only on the training dataset to prevent data leakage.

The final datasets were checked for missing and infinite values.
